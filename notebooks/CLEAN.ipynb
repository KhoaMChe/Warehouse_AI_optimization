## Vị trí

```python
print(f"Shape: {vitri.shape}")
print(f"info: {vitri.info()}")
```

```python
print(f"\nMissing values:\n{vitri.isnull().sum().sort_values(ascending=False).head(50)}")
```

```python
#drop những hàng và cột data không cần thiết và missing value quá nhiều
drop_colums = ["tag","ghi_chu"]
vitri = vitri.drop(columns=drop_colums)
```

```python
# Xóa dòng trùng
vitri = vitri.drop_duplicates()
```

```python
#kiểm tra có bao nhiêu loại tầng
vitri["tang"].nunique()
```

```python
#những tầng thực tế cụ thể
vitri["tang"].unique()
```

```python
#số lượng vị trí các tầng có đầy đủ thông số dài, rộng, cao
vitri.groupby("tang")[["dai", "rong", "cao"]].apply(
    lambda x: x.notna().all(axis=1).sum()
)
```

```python
#chuyển 0 thành NaN để không ảnh hưởng đến tính gtri trung bình
for col in ["dai", "rong", "cao"]:
    vitri[col] = vitri[col].replace(0, np.nan)
#tính trung bình
for col in ["dai", "rong", "cao"]:
    vitri[col] = vitri[col].mask(
        vitri[col] == 0,
        vitri.groupby("tang")[col].transform("mean")
    )

# Điền chiều dài theo trung bình của cùng tầng
vitri["dai"] = vitri["dai"].fillna(
    vitri.groupby("tang")["dai"].transform("mean")
)

# Điền chiều rộng
vitri["rong"] = vitri["rong"].fillna(
    vitri.groupby("tang")["rong"].transform("mean")
)

# Điền chiều cao
vitri["cao"] = vitri["cao"].fillna(
    vitri.groupby("tang")["cao"].transform("mean")
)
```

```python
#lấy cbm max của các tầng 7,8,9,10 
vitri[vitri["tang"].isin([7, 8, 9, 10])] \
    .groupby("tang")["cbm_max"] \
    .unique()
```

```python
vitri[
    vitri["cbm_max"].isin([1782500, 1860000])
].groupby("tang")["cbm_max"].unique()
```

```python
cols = ["dai", "rong", "cao"]

# Chuyển 0 thành NaN để không ảnh hưởng khi tính trung bình
vitri[cols] = vitri[cols].replace(0, np.nan)

# Điền theo cbm_max
for col in cols:
    vitri[col] = vitri[col].fillna(
        vitri.groupby("cbm_max")[col].transform("mean")
    )
```

```python
# Đổi 0 thành NaN để không ảnh hưởng đến trung bình
vitri["cbm_max"] = vitri["cbm_max"].replace(0, np.nan)

# Điền cbm_max theo trung bình của cùng tầng
vitri["cbm_max"] = vitri["cbm_max"].fillna(
    vitri.groupby("tang")["cbm_max"].transform("mean")
)
```
