## Tỉ lệ lấp đầy kho

```python
agg = tonkho.groupby("vi_tri_id").agg(
    so_sku=("san_pham_id", "nunique"),
    so_lpn=("so_lpn", "nunique"),
    tong_sl=("sl_nhap_chan", "sum")  # hoặc sl tồn bạn đang dùng
).reset_index()

df = vitri.merge(
    agg,
    left_on="auto_id",
    right_on="vi_tri_id",
    how="left"
)

df["so_sku"] = df["so_sku"].fillna(0)
df["so_lpn"] = df["so_lpn"].fillna(0)

df["trang_thai_vi_tri"] = df["so_sku"].apply(
    lambda x: "TRONG" if x == 0 else "CO_HANG"
)

cohang = (df["trang_thai_vi_tri"] == "CO_HANG").sum()
trong = (df["trang_thai_vi_tri"] == "TRONG").sum()
tong = len(df)
```

```python
fig,ax=plt.subplots(1,2,figsize=(11,4))

# Bar

ax[0].bar(
    ["Có Hàng","Trống"],
    [cohang,trong],
    color=["#59139A","#E35825"]
)

for i,v in enumerate([cohang,trong]):
    ax[0].text(
        i,
        v+20,
        f"{v:,}\n({v/tong:.1%})",
        ha="center",
        fontweight="bold"
    )

ax[0].set_ylabel("Số lượng vị trí")
ax[0].set_title("Hàng có trong kho")

# Pie

ax[1].pie(
    [cohang,trong],
    labels=["Có hàng","Trống"],
    autopct="%1.1f%%",
    colors=["#59139A","#E35825"],
    startangle=90
)

ax[1].set_title(
    f"Tỉ lệ có hàng = {cohang/tong:.1%}"
)

plt.tight_layout()

plt.savefig(IMAGES/"warehouse_occupancy.png",dpi=150)

plt.show()

print(
    f"Tỉ lệ có hàng = {cohang/tong:.1%}"
)
```

## Quan sát Rules của kho

```python
# Ghép tồn kho với thông tin vị trí
sku_rule = (
    tonkho.merge(
        vitri[
            ["auto_id", "day_ke_id", "tang"]
        ],
        left_on="vi_tri_id",
        right_on="auto_id",
        how="left"
    )
)

```

```python
#lấy số dãy
so_day = (
    sku_rule
    .groupby(["kho_id","san_pham_id"])["day_ke_id"]
    .nunique()
    .rename("so_day")
)
```

```python
#tính toán các thông số cần thiết
day_rule = (
    sku_rule
    .groupby(
        ["kho_id", "san_pham_id", "day_ke_id"]
    )["vi_tri_id"]
    .nunique()
    .reset_index(name="so_vi_tri")
)
day_rule["ratio"] = (
    day_rule["so_vi_tri"]
    /
    day_rule.groupby(
        ["kho_id", "san_pham_id"]
    )["so_vi_tri"].transform("sum")
)

day_chinh = (
    day_rule
    .sort_values(
        ["kho_id","san_pham_id","ratio"],
        ascending=[True,True,False]
    )
    .drop_duplicates(["kho_id","san_pham_id"])
    [["kho_id","san_pham_id","day_ke_id","ratio"]]
)

day_chinh.columns = [
    "kho_id",
    "san_pham_id",
    "day_chinh",
    "ty_le"
]
```

```python
so_tang = (
    sku_rule
    .groupby(["kho_id","san_pham_id"])["tang"]
    .nunique()
    .rename("so_tang")
)
```

```python
floor_rule = (
    sku_rule
    .groupby(
        ["kho_id", "san_pham_id", "tang"]
    )["vi_tri_id"]
    .nunique()
    .reset_index(name="so_vi_tri")
)

tang_chinh = (
    floor_rule
    .sort_values(
        ["kho_id", "san_pham_id", "so_vi_tri"],
        ascending=[True, True, False]
    )
    .drop_duplicates(
        ["kho_id", "san_pham_id"]
    )
    .rename(
        columns={
            "tang": "tang_chinh"
        }
    )[
        [
            "kho_id",
            "san_pham_id",
            "tang_chinh"
        ]
    ]
)
```

```python
so_tang = (
    sku_rule
    .groupby(
        ["kho_id", "san_pham_id"]
    )["tang"]
    .nunique()
    .reset_index(name="so_tang")
)
```

```python
summary = (
    day_chinh
    .merge(
        so_day,
        on=["kho_id","san_pham_id"]
    )
    .merge(
        so_tang,
        on=["kho_id","san_pham_id"]
    )
    .merge(
        tang_chinh,
        on=["kho_id","san_pham_id"]
    )
)

summary["ty_le"] = (summary["ty_le"]*100).round(2)

summary = summary[
    [
        "kho_id",
        "san_pham_id",
        "so_day",
        "day_chinh",
        "ty_le",
        "so_tang",
        "tang_chinh"
    ]
]

summary.head(20)
```

```python
#tính toán các thông số cần thiết
day_rule = (
    rule_cham
    .groupby(
        ["kho_id", "san_pham_id", "day_ke_id"]
    )["auto_id_y"]
    .nunique()
    .reset_index(name="so_vi_tri")
)
day_rule["ratio"] = (
    day_rule["so_vi_tri"]
    /
    day_rule.groupby(
        ["kho_id", "san_pham_id"]
    )["so_vi_tri"].transform("sum")
)

day_chinh = (
    day_rule
    .sort_values(
        ["kho_id","san_pham_id","ratio"],
        ascending=[True,True,False]
    )
    .drop_duplicates(["kho_id","san_pham_id"])
    [["kho_id","san_pham_id","day_ke_id","ratio"]]
)

day_chinh.columns = [
    "kho_id",
    "san_pham_id",
    "day_chinh",
    "ty_le"
]
```

```python
so_tang_cham = (
    rule_cham
    .groupby(["kho_id","san_pham_id"])["tang"]
    .nunique()
    .rename("so_tang")
)
```

```python
floor_rule = (
    rule_cham
    .groupby(
        ["kho_id", "san_pham_id", "tang"]
    )["auto_id_y"]
    .nunique()
    .reset_index(name="so_vi_tri")
)

tang_chinh = (
    floor_rule
    .sort_values(
        ["kho_id", "san_pham_id", "so_vi_tri"],
        ascending=[True, True, False]
    )
    .drop_duplicates(
        ["kho_id", "san_pham_id"]
    )
    .rename(
        columns={
            "tang": "tang_chinh"
        }
    )[
        [
            "kho_id",
            "san_pham_id",
            "tang_chinh"
        ]
    ]
)
```

```python
so_tang = (
    rule_cham
    .groupby(
        ["kho_id", "san_pham_id"]
    )["tang"]
    .nunique()
    .reset_index(name="so_tang")
)
```

```python
summary1 = (
    day_chinh
    .merge(
        so_day_cham,
        on=["kho_id","san_pham_id"]
    )
    .merge(
        so_tang,
        on=["kho_id","san_pham_id"]
    )
    .merge(
        tang_chinh,
        on=["kho_id","san_pham_id"]
    )
)

summary1["ty_le"] = (summary1["ty_le"]*100).round(2)

summary1 = summary1[
    [
        "kho_id",
        "san_pham_id",
        "so_day",
        "day_chinh",
        "ty_le",
        "so_tang",
        "tang_chinh"
    ]
]
summary1["so_day"] = summary1["so_day"].apply(len)
summary1.head(20)
```
