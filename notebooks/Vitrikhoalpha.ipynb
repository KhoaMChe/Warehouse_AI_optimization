```python
active_sku = set(tonkho["san_pham_id"].unique())

active_sku |= set(xuatkho["san_pham_id"].dropna().unique())

active_sku |= set(nhapkho["san_pham_id"].dropna().unique())

active_sku |= set(cham["san_pham_id"].dropna().unique())

active_sku = pd.Series(list(active_sku))

print(len(active_sku))
```

```python
KHO_ID = 12473657

active_sku = set(
    tonkho.loc[
        tonkho["kho_id"] == KHO_ID,
        "san_pham_id"
    ].dropna().unique()
)

active_sku |= set(
    xuatkho.loc[
        xuatkho["kho_id"] == KHO_ID,
        "san_pham_id"
    ].dropna().unique()
)

# Chỉ thêm nếu bảng châm hàng thật sự có san_pham_id và kho_id
active_sku |= set(
    cham.loc[
        cham["kho_id"] == KHO_ID,
        "san_pham_id"
    ].dropna().unique()
)

SPkhoA = (
    pd.Series(sorted(active_sku), name="san_pham_id")
)

print(len(SPkhoA))
SPkhoA
```

```python
active_df = pd.DataFrame({
    "san_pham_id": list(active_sku)
})

SPkhoA = (
    active_df.merge(
        sanpham,
        left_on="san_pham_id",
        right_on="auto_id",
        how="left"
    )
    .sort_values(
        ["nganh_hang_id","loai_san_pham_id", "ma_san_pham"]
    )
    .reset_index(drop=True)
)

SPkhoA
```

```python
SPkhoA.to_csv(
    "../data/clean/12473657/san_pham_kho12473657.csv",
    index=False,
    encoding="utf-8-sig"
)
```
